# Clusters con DBSCAN y Scikit-learn

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

# Cargar datos
datos = np.loadtxt('casas.csv', delimiter=',')

# Identificar Clusters
agrupacion = DBSCAN(eps=2, min_samples=10).fit_predict(datos)
print(agrupacion)

# Gráfica de matplotlib para mostrar los Clusters
plt.figure(figsize=(7.5, 7.5))
plt.scatter(datos[:, 0], datos[:, 1], c=agrupacion, s=100)
plt.xlabel('Antigüedad del inmueble (en años)')
plt.ylabel('Precio de la casa (en pesos) (1:100,000)')
plt.box(False)
plt.show()
```
